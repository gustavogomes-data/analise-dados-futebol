```python
df_bruto = pd.read_csv("C:/Users/Gustavo/Documents/DadosFut/DadosFUTBrutos.csv", header=[0, 1]) # leio o arquivo bruto que foi gerado, utilizando header=[0,1] para manter a estrutura de multi-index do cabeçalho
df_bruto.head(5)
```

```python
df_bruto.columns

#Verifiquei que Temos um MultiIndex no cabeçalho, então vamos renomear as colunas para facilitar a manipulação dos dados.
```

```python
df_bruto.columns = df_bruto.columns.map('_'.join) #vou juntar o nivel 1 e 0 e transformar em uma unica coluna, separando por _
```

```python
df_bruto.columns 
#colunas juntas, agora vamos renomear 
```

```python
df_bruto = df_bruto.rename(columns={
    'league_Unnamed: 0_level_1': 'league',
    'season_Unnamed: 1_level_1': 'season',
    'team_Unnamed: 2_level_1': 'team',
    'player_Unnamed: 3_level_1': 'player',
    'nation_Unnamed: 4_level_1': 'nation',
    'pos_Unnamed: 5_level_1': 'pos',
    'age_Unnamed: 6_level_1': 'age',
    'born_Unnamed: 7_level_1': 'born',

    'Playing Time_MP': 'MP',
    'Playing Time_Starts': 'Starts',
    'Playing Time_Min': 'Min',
    'Playing Time_90s': '90s',

    'Performance_Gls': 'Gls',
    'Performance_Ast': 'Ast',
    'Performance_G+A': 'G+A',
    'Performance_G-PK': 'G-PK',

    'Per 90 Minutes_Gls': 'Gls_90',
    'Per 90 Minutes_Ast': 'Ast_90',
    'Per 90 Minutes_G+A': 'G+A_90',
    'Per 90 Minutes_G-PK': 'G-PK_90',
    'Per 90 Minutes_G+A-PK': 'G+A-PK_90'
})

#colunas renomeadas, agora vamos selecionar apenas as colunas que queremos manter
```

```python
df_bruto = df_bruto[
    [
        'league',
        'season',
        'team',
        'player',
        'nation',
        'pos',
        'age',
        'MP',
        'Starts',
        'Min',
        'Gls',
        'Ast',
        'G+A',
        'G-PK',
        'Gls_90',
        'Ast_90',
        'G+A_90'
    ]
]

#selecionada colunas que vamos utilizar no projeto.
```

```python

df_bruto['pos'].value_counts() #verifiquei a quantidade de jogadores por posição, e percebi que temos jogadores de todas as posições, então vamos selecionar apenas os jogadores que são meio-campistas ou atacantes, pois são os que mais contribuem para o ataque do time.
```

```python
df_bruto = df_bruto[df_bruto['pos'].str.contains('MF|FW')] #Seleciono apenas os jogadores que são meio-campistas ou atacantes, pois são os que mais contribuem para o ataque do time.
```

```python
df_bruto.duplicated().sum() #verifico se tem valores duplicados, caso tenha, vou remover, pois não faz sentido ter jogadores duplicados no mesmo time.
```

```python
df_bruto = df_bruto[df_bruto['Min'] > 900] #seleciono apenas os jogadores que jogaram mais de 900 minutos, pois são os que mais contribuíram para o time, e são os que mais tem chances de serem titulares.
```
